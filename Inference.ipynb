```python
import os
import shutil
import easyocr
import joblib
import re

# ----------------------
# Paths
# ----------------------
INPUT_FOLDER = "/content/drive/MyDrive/project/data"  # Folder containing images to classify
OUTPUT_FOLDER = "/content/drive/MyDrive/project/Segregated_Data"       # Root folder where images will be moved

# ----------------------
# Load trained model and vectorizer
# ----------------------
model = joblib.load("/content/drive/MyDrive/project/Source_code/model.pkl")
vectorizer = joblib.load("/content/drive/MyDrive/project/Source_code/vectorizer.pkl")

# ----------------------
# Initialize OCR
# ----------------------
reader = easyocr.Reader(['en'])

# ----------------------
# Class labels (folder names)
# ----------------------
class_labels = ['letter', 'resume', 'scientific']

# Ensure all class folders exist
for label in class_labels:
    os.makedirs(os.path.join(OUTPUT_FOLDER, label), exist_ok=True)

# ----------------------
# OCR + cleaning function
# ----------------------
def extract_text(image_path):
    try:
        result = reader.readtext(image_path, detail=0)
        text = " ".join(result)
        text = re.sub(r'[^A-Za-z0-9\s]+', ' ', text)  # remove special characters
        return text.lower()
    except Exception as e:
        print(f"Error reading {image_path}: {e}")
        return ""

# ----------------------
# Prediction function
# ----------------------
def predict_label(image_path):
    text = extract_text(image_path)
    if text.strip() == "":
        return None  # skip if no text detected
    features = vectorizer.transform([text])
    prediction = model.predict(features)
    return prediction[0]

# ----------------------
# Process all images
# ----------------------
summary = {label: 0 for label in class_labels}

for filename in os.listdir(INPUT_FOLDER):
    file_path = os.path.join(INPUT_FOLDER, filename)

    # Only process image files
    if not filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        continue

    label = predict_label(file_path)

    if label is None:
        print(f"Skipped {filename} (no text found)")
        continue

    # Ensure destination folder exists
    dest_folder = os.path.join(OUTPUT_FOLDER, label)
    os.makedirs(dest_folder, exist_ok=True)

    # Move file to corresponding folder
    dest_path = os.path.join(dest_folder, filename)
    try:
        shutil.move(file_path, dest_path)
        summary[label] += 1
        print(f"{filename} → {label}")
    except Exception as e:
        print(f"Error moving {filename}: {e}")

# ----------------------
# Print summary
# ----------------------
print("\n✅ All images processed and segregated successfully!")
print("📊 Summary:")
for label, count in summary.items():
    print(f"{label}: {count} images")
print("done")
```
